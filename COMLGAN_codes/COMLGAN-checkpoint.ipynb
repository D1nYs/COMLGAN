```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader
import os
import glob
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import cv2
import random
```

```python
class SimpleUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super(SimpleUNet, self).__init__()
        
        # Encoder
        self.enc1 = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        self.pool1 = nn.MaxPool2d(2)
        
        self.enc2 = nn.Sequential(
            nn.Conv2d(16, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.pool2 = nn.MaxPool2d(2)
        
        # Bottleneck
        self.bottleneck = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU()
        )
        
        # Decoder
        self.up2 = nn.ConvTranspose2d(64, 32, 2, stride=2)
        self.dec2 = nn.Sequential(
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        
        self.up1 = nn.ConvTranspose2d(32, 16, 2, stride=2)
        self.dec1 = nn.Sequential(
            nn.Conv2d(32, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        
        # Output
        self.out = nn.Conv2d(16, 3, 1)
        
    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        p1 = self.pool1(e1)
        
        e2 = self.enc2(p1)
        p2 = self.pool2(e2)
        
        # Bottleneck
        b = self.bottleneck(p2)
        
        # Decoder
        u2 = self.up2(b)
        # Resize if needed
        if u2.shape != e2.shape:
            u2 = torch.nn.functional.interpolate(u2, size=e2.shape[2:], mode='bilinear')
        d2 = self.dec2(torch.cat([u2, e2], dim=1))
        
        u1 = self.up1(d2)
        if u1.shape != e1.shape:
            u1 = torch.nn.functional.interpolate(u1, size=e1.shape[2:], mode='bilinear')
        d1 = self.dec1(torch.cat([u1, e1], dim=1))
        
        return self.out(d1)


```

```python
class Generator(nn.Module):
    def __init__(self, latent_dim=100, num_classes=5, img_channels=3):
        super(Generator, self).__init__()
        self.num_classes = num_classes
        self.latent_dim = latent_dim
        
        self.init_size = 16
        self.label_embedding = nn.Embedding(num_classes, latent_dim)
        
        self.l1 = nn.Sequential(nn.Linear(latent_dim * 2, 128 * self.init_size ** 2))
        
        self.conv_blocks = nn.Sequential(
            nn.BatchNorm2d(128),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 128, 3, stride=1, padding=1),
            nn.BatchNorm2d(128, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout2d(0.1),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 64, 3, stride=1, padding=1),
            nn.BatchNorm2d(64, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, img_channels, 3, stride=1, padding=1),
            nn.Tanh()
        )
    
    def forward(self, z, labels):
        label_embed = self.label_embedding(labels)
        gen_input = torch.cat((label_embed, z), -1)
        
        out = self.l1(gen_input)
        out = out.view(out.shape[0], 128, self.init_size, self.init_size)
        img = self.conv_blocks(out)
        return img

class Discriminator(nn.Module):
    def __init__(self, num_classes=5, img_channels=3):
        super(Discriminator, self).__init__()
        
        def discriminator_block(in_filters, out_filters, bn=True):
            block = [nn.Conv2d(in_filters, out_filters, 3, 2, 1),
                    nn.LeakyReLU(0.2, inplace=True),
                    nn.Dropout2d(0.3)]
            if bn:
                block.append(nn.BatchNorm2d(out_filters, 0.8))
            return block
        
        self.model = nn.Sequential(
            *discriminator_block(img_channels, 16, bn=False),
            *discriminator_block(16, 32),
            *discriminator_block(32, 64),
            *discriminator_block(64, 128),
        )
        
        ds_size = 64 // 2 ** 4
        self.adv_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, 1), nn.Sigmoid())
        self.aux_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, num_classes), nn.Softmax(dim=1))
    
    def forward(self, img):
        out = self.model(img)
        out = out.view(out.shape[0], -1)
        validity = self.adv_layer(out)
        label = self.aux_layer(out)
        return validity, label


```

```python
class ACGANDataset(Dataset):
    def __init__(self, normal_folder, defect_folders, transform=None, img_size=64):
        self.transform = transform
        self.img_size = img_size
        self.image_paths = []
        self.labels = []
        
        self.class_names = ['normal', 'abrasive', 'adhesive', 'cracks', 'delamination']
        
        normal_images = self._get_image_paths(normal_folder)
        self.image_paths.extend(normal_images)
        self.labels.extend([0] * len(normal_images))
        
        for i, defect_folder in enumerate(defect_folders, 1):
            defect_images = self._get_image_paths(defect_folder)
            self.image_paths.extend(defect_images)
            self.labels.extend([i] * len(defect_images))
        
        print(f"Loaded {len(self.image_paths)} images")
        for i, class_name in enumerate(self.class_names):
            count = sum(1 for label in self.labels if label == i)
            print(f"Class {class_name}: {count} images")
    
    def _get_image_paths(self, folder_path):
        extensions = ['*.jpg', '*.jpeg', '*.png', '*.bmp', '*.JPG', '*.JPEG', '*.PNG', '*.BMP']
        image_paths = []
        
        for extension in extensions:
            image_paths.extend(glob.glob(os.path.join(folder_path, extension)))
        
        return image_paths
        
    def __len__(self):
        return len(self.image_paths)
        
    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        label = self.labels[idx]
        
        try:
            image = Image.open(img_path).convert('RGB')
            
            if self.transform:
                image = self.transform(image)
            else:
                transform = transforms.Compose([
                    transforms.Resize((self.img_size, self.img_size)),
                    transforms.ToTensor(),
                    transforms.Normalize([0.5] * 3, [0.5] * 3)
                ])
                image = transform(image)
                
            return image, label
            
        except Exception as e:
            print(f"Error loading image {img_path}: {e}")
            dummy_image = torch.zeros(3, self.img_size, self.img_size)
            return dummy_image, label


```

```python
class ACGANAnomalyDetection:
    def __init__(self, num_classes=5, img_size=64, channels=3, latent_dim=100):
        self.num_classes = num_classes
        self.img_size = img_size
        self.channels = channels
        self.latent_dim = latent_dim
        
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"Using device: {self.device}")

        self.unet = SimpleUNet(in_channels=3, out_channels=1).to(self.device)
        self.generator = Generator(latent_dim, num_classes, channels).to(self.device)
        self.discriminator = Discriminator(num_classes, channels).to(self.device)
        
        # Load pretrained U-Net if available
        if os.path.exists(r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\2_unet_trained.pth"):
            self.unet.load_state_dict(torch.load(r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\2_unet_trained.pth", map_location=self.device))
        
        # Loss functions
        self.adversarial_loss = nn.BCELoss()
        self.auxiliary_loss = nn.CrossEntropyLoss()
        self.unet_loss = nn.BCEWithLogitsLoss()
        
        # Optimisers
        self.optimizer_G = optim.Adam(self.generator.parameters(), lr=0.0002, betas=(0.5, 0.999))
        self.optimizer_D = optim.Adam(self.discriminator.parameters(), lr=0.0002, betas=(0.5, 0.999))
        self.optimizer_U = optim.Adam(self.unet.parameters(), lr=0.0001, betas=(0.5, 0.999))  # Меньше LR для UNet
        
        # Transformers
        self.transform = transforms.Compose([
            transforms.Resize((img_size, img_size)),
            transforms.ToTensor(),
            transforms.Normalize([0.5] * channels, [0.5] * channels)
        ])
        
        self.class_names = ['normal', 'abrasive', 'adhesive', 'cracks', 'delamination']

        # Visualisation
        self.fixed_z = torch.randn(num_classes, latent_dim).to(self.device)
        self.fixed_labels = torch.arange(num_classes).to(self.device)
    
    def train(self, dataloader, epochs):
        """Concurrent training of U-Net and GAN"""
        self.generator.train()
        self.discriminator.train()
        self.unet.train()  # IMPORTANT: U-Net in training mode
        
        for epoch in range(epochs):
            epoch_g_loss = 0
            epoch_d_loss = 0
            epoch_u_loss = 0
            
            for i, (imgs, labels) in enumerate(dataloader):
                batch_size = imgs.size(0)
                real_imgs = imgs.to(self.device)
                real_labels = labels.to(self.device)
                
                valid = torch.ones(batch_size, 1, device=self.device)
                fake = torch.zeros(batch_size, 1, device=self.device)
                
                # ===== STEP 1: TRAINING THE GENERATOR AND U-NET =====
                self.optimizer_G.zero_grad()
                self.optimizer_U.zero_grad()
                
                # Image generation
                gen_labels = torch.randint(0, self.num_classes, (batch_size,), device=self.device)
                z = torch.randn(batch_size, self.latent_dim, device=self.device)
                gen_imgs = self.generator(z, gen_labels)
                
                # U-Net prediction
                unet_mask = self.unet(real_imgs)
                
                # Generator loss
                validity, pred_label = self.discriminator(gen_imgs)
                g_loss_adv = self.adversarial_loss(validity, valid)
                g_loss_aux = self.auxiliary_loss(pred_label, gen_labels) * 2.0
                
                with torch.no_grad():
                    gan_diff = torch.abs(real_imgs - gen_imgs).mean(dim=1, keepdim=True)
                    pseudo_gt = (gan_diff > 0.3).float()
                    
                    pseudo_gt = pseudo_gt.repeat(1, 3, 1, 1)
                    
                u_loss = self.unet_loss(unet_mask, pseudo_gt) * 0.2
                
                # Total loss for generator and U-Net
                total_g_u_loss = g_loss_adv + g_loss_aux + u_loss
                total_g_u_loss.backward()
                self.optimizer_G.step()
                self.optimizer_U.step()
                
                self.optimizer_D.zero_grad()
                
                #  Loss on real images
                real_validity, real_pred_label = self.discriminator(real_imgs)
                d_real_loss_adv = self.adversarial_loss(real_validity, valid)
                d_real_loss_aux = self.auxiliary_loss(real_pred_label, real_labels)
                
                #Loss on generated images
                fake_validity, _ = self.discriminator(gen_imgs.detach())
                d_fake_loss_adv = self.adversarial_loss(fake_validity, fake)
                
                d_loss_adv = (d_real_loss_adv + d_fake_loss_adv) / 2
                d_loss_aux = d_real_loss_aux
                d_loss = d_loss_adv + d_loss_aux
                
                d_loss.backward()
                self.optimizer_D.step()
                
                # Collect statistics
                epoch_g_loss += total_g_u_loss.item()
                epoch_d_loss += d_loss.item()
                epoch_u_loss += u_loss.item()
                
                # Logs
                if i % 50 == 0:
                    print(f"[Epoch {epoch+1}/{epochs}] [Batch {i}/{len(dataloader)}] "
                          f"[D loss: {d_loss.item():.4f}] [G loss: {total_g_u_loss.item():.4f}] "
                          f"[UNet loss: {u_loss.item():.4f}]")
            
            # Average loss per epoch
            avg_g_loss = epoch_g_loss / len(dataloader)
            avg_d_loss = epoch_d_loss / len(dataloader)
            avg_u_loss = epoch_u_loss / len(dataloader)
    
    def visualize_training_progress(self, dataloader, epoch):
        """Visualize training progress"""
        self.generator.eval()
        self.discriminator.eval()
        self.unet.eval()
        
        # Take one batch for visualization
        for batch in dataloader:
            real_imgs, _ = batch
            real_imgs = real_imgs[:self.num_classes].to(self.device)
            break
        
        with torch.no_grad():
            # Generating images for each class
            gen_imgs = self.generator(self.fixed_z, self.fixed_labels)
            
            # UNet masks
            unet_masks = self.unet(real_imgs)
            
            fig, axes = plt.subplots(4, self.num_classes, figsize=(15, 10))
            
            for i in range(self.num_classes):
                # 1. Generated images
                gen_img = gen_imgs[i].cpu().permute(1, 2, 0).numpy()
                gen_img = (gen_img * 0.5 + 0.5)  # Denormalize [0, 1]
                axes[0, i].imshow(np.clip(gen_img, 0, 1))
                axes[0, i].set_title(f'Gen: {self.class_names[i]}')
                axes[0, i].axis('off')
                
                # 2. UNet masks
                if i < len(unet_masks):
                    unet_mask = unet_masks[i].cpu().squeeze().numpy()
                    axes[1, i].imshow(unet_mask, cmap='hot')
                    axes[1, i].set_title('UNet Mask')
                    axes[1, i].axis('off')
                
                # 3. GAN reconstructions
                z = torch.randn(1, self.latent_dim, device=self.device)
                label = torch.tensor([i], device=self.device)
                recon = self.generator(z, label)
                
                if i < len(real_imgs):
                    gan_diff = torch.abs(real_imgs[i] - recon[0]).mean(0).cpu().numpy()
                    axes[2, i].imshow(gan_diff, cmap='hot')
                    axes[2, i].set_title('GAN Diff')
                    axes[2, i].axis('off')
                
                # 4. Original images
                if i < len(real_imgs):
                    real_img = real_imgs[i].cpu().permute(1, 2, 0).numpy()
                    real_img = (real_img * 0.5 + 0.5)  # Denormalize
                    axes[3, i].imshow(np.clip(real_img, 0, 1))
                    axes[3, i].set_title('Original')
                    axes[3, i].axis('off')
        
        self.generator.train()
        self.discriminator.train()
        self.unet.train()
    
    def detect_anomalies(self, image_path, threshold=0.5):
        """Image anomaly detection"""
        self.generator.eval()
        self.discriminator.eval()
        self.unet.eval()
        
        try:
            image = Image.open(image_path).convert('RGB')
            original_size = image.size
            image_tensor = self.transform(image).unsqueeze(0).to(self.device)
            
            with torch.no_grad():
                # Classifications
                validity, class_probs = self.discriminator(image_tensor)
                predicted_class = torch.argmax(class_probs, dim=1).item()
                confidence = class_probs[0, predicted_class].item()
                
                # Reconstruction GAN
                z = torch.randn(1, self.latent_dim).to(self.device)
                label_tensor = torch.tensor([predicted_class], device=self.device)
                reconstructed = self.generator(z, label_tensor)
                
                # UNet mask
                unet_mask = self.unet(image_tensor)
                unet_mask = unet_mask.mean(dim=1, keepdim=True)
                
                # GAN difference
                gan_difference = torch.abs(image_tensor - reconstructed)
                
                # Combined U-Net and GAN
                combined_anomaly = (unet_mask + gan_difference.mean(dim=1, keepdim=True)) / 2
                
                # Thresholding
                anomaly_mask = (combined_anomaly > threshold).float()
                
                # Output data preparation
                anomaly_mask_np = anomaly_mask.squeeze().cpu().numpy()
                anomaly_mask_resized = cv2.resize(anomaly_mask_np, original_size)
                
                return {
                    'anomaly_mask': anomaly_mask_resized,
                    'predicted_class': predicted_class,
                    'class_name': self.class_names[predicted_class],
                    'confidence': confidence,
                    'class_probs': class_probs.cpu().numpy(),
                    'difference': combined_anomaly.squeeze().cpu().numpy(),
                    'unet_mask': unet_mask.squeeze().cpu().numpy(),
                    'gan_difference': gan_difference.squeeze().cpu().numpy()
                }
                
        except Exception as e:
            print(f"Anomaly detection error: {e}")
            return None       
    
    def visualize_results(self, image_paths):
        """Visualize results"""
        anomaly_colors = {
            0: [0, 255, 0],    # Green - normal
            1: [255, 0, 0],    # Red - abrasive
            2: [0, 0, 255],    # Blue - adhesive
            3: [255, 255, 0],  # Yellow - cracks
            4: [255, 0, 255]   # purple - delamination
        }
        
        for i, img_path in enumerate(image_paths):
            print(f"\nImage processing {i+1}/{len(image_paths)}: {os.path.basename(img_path)}")
            
            result = self.detect_anomalies(img_path, threshold=0.5)
            if result is None:
                continue
            
            # Load original image
            original_image = cv2.imread(img_path)
            if original_image is None:
                continue
            
            original_image_rgb = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
            
            # Resize mask to original image size
            anomaly_mask_resized = cv2.resize(
                result['anomaly_mask'], 
                (original_image_rgb.shape[1], original_image_rgb.shape[0])
            )
            if len(anomaly_mask_resized.shape) == 3:
                anomaly_mask_resized = anomaly_mask_resized.mean(axis=2)
                
            # Color mask
            color_mask = np.zeros_like(original_image_rgb)
            color_mask[anomaly_mask_resized > 0] = anomaly_colors[result['predicted_class']]
            
            # Overlay
            result_image = cv2.addWeighted(original_image_rgb, 0.7, color_mask, 0.3, 0)
            
            # Visualisation
            plt.figure(figsize=(20, 5))
            
            plt.subplot(1, 4, 1)
            plt.imshow(original_image_rgb)
            plt.title('Original Image')
            plt.axis('off')
            
            plt.subplot(1, 4, 2)
            diff_visual = result['difference'].mean(axis=0) if len(result['difference'].shape) == 3 else result['difference']
            plt.imshow(diff_visual, cmap='hot')
            plt.title('Difference Map')
            plt.axis('off')
            
            plt.subplot(1, 4, 3)
            plt.imshow(anomaly_mask_resized, cmap='gray')
            plt.title('Anomaly Mask')
            plt.axis('off')
            
            plt.subplot(1, 4, 4)
            plt.imshow(result_image)
            plt.title(f'Class: {result["class_name"]}\nConfidence: {result["confidence"]:.3f}')
            plt.axis('off')
            
            plt.tight_layout()
            plt.show()
            
            print("Class probabilities:")
            for j, class_name in enumerate(self.class_names):
                prob = result['class_probs'][0, j]
                print(f"  {class_name}: {prob:.4f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader
import os
import glob
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import cv2
import random
```

```python
class SimpleUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super(SimpleUNet, self).__init__()
        
        # Encoder
        self.enc1 = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        self.pool1 = nn.MaxPool2d(2)
        
        self.enc2 = nn.Sequential(
            nn.Conv2d(16, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.pool2 = nn.MaxPool2d(2)
        
        # Bottleneck
        self.bottleneck = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU()
        )
        
        # Decoder
        self.up2 = nn.ConvTranspose2d(64, 32, 2, stride=2)
        self.dec2 = nn.Sequential(
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        
        self.up1 = nn.ConvTranspose2d(32, 16, 2, stride=2)
        self.dec1 = nn.Sequential(
            nn.Conv2d(32, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        
        # Output
        self.out = nn.Conv2d(16, 3, 1)
        
    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        p1 = self.pool1(e1)
        
        e2 = self.enc2(p1)
        p2 = self.pool2(e2)
        
        # Bottleneck
        b = self.bottleneck(p2)
        
        # Decoder
        u2 = self.up2(b)
        # Resize if needed
        if u2.shape != e2.shape:
            u2 = torch.nn.functional.interpolate(u2, size=e2.shape[2:], mode='bilinear')
        d2 = self.dec2(torch.cat([u2, e2], dim=1))
        
        u1 = self.up1(d2)
        if u1.shape != e1.shape:
            u1 = torch.nn.functional.interpolate(u1, size=e1.shape[2:], mode='bilinear')
        d1 = self.dec1(torch.cat([u1, e1], dim=1))
        
        return self.out(d1)


```

```python
class Generator(nn.Module):
    def __init__(self, latent_dim=100, num_classes=5, img_channels=3):
        super(Generator, self).__init__()
        self.num_classes = num_classes
        self.latent_dim = latent_dim
        
        self.init_size = 16
        self.label_embedding = nn.Embedding(num_classes, latent_dim)
        
        self.l1 = nn.Sequential(nn.Linear(latent_dim * 2, 128 * self.init_size ** 2))
        
        self.conv_blocks = nn.Sequential(
            nn.BatchNorm2d(128),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 128, 3, stride=1, padding=1),
            nn.BatchNorm2d(128, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout2d(0.1),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 64, 3, stride=1, padding=1),
            nn.BatchNorm2d(64, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, img_channels, 3, stride=1, padding=1),
            nn.Tanh()
        )
    
    def forward(self, z, labels):
        label_embed = self.label_embedding(labels)
        gen_input = torch.cat((label_embed, z), -1)
        
        out = self.l1(gen_input)
        out = out.view(out.shape[0], 128, self.init_size, self.init_size)
        img = self.conv_blocks(out)
        return img

class Discriminator(nn.Module):
    def __init__(self, num_classes=5, img_channels=3):
        super(Discriminator, self).__init__()
        
        def discriminator_block(in_filters, out_filters, bn=True):
            block = [nn.Conv2d(in_filters, out_filters, 3, 2, 1),
                    nn.LeakyReLU(0.2, inplace=True),
                    nn.Dropout2d(0.3)]
            if bn:
                block.append(nn.BatchNorm2d(out_filters, 0.8))
            return block
        
        self.model = nn.Sequential(
            *discriminator_block(img_channels, 16, bn=False),
            *discriminator_block(16, 32),
            *discriminator_block(32, 64),
            *discriminator_block(64, 128),
        )
        
        ds_size = 64 // 2 ** 4
        self.adv_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, 1), nn.Sigmoid())
        self.aux_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, num_classes), nn.Softmax(dim=1))
    
    def forward(self, img):
        out = self.model(img)
        out = out.view(out.shape[0], -1)
        validity = self.adv_layer(out)
        label = self.aux_layer(out)
        return validity, label


```

```python
def main():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Используется устройство: {device}")
    
    gan_model_path = r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\final_hybrid_model.pth"
    unet_model_path = r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\2_unet_trained.pth"
    
    
    image_path = r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\tzh_wear\delamination\pic1_4_16.png"
    
    
    if not os.path.exists(gan_model_path):
        print(f"Файл GAN модели не найден: {gan_model_path}")
        return
    
    if not os.path.exists(image_path):
        print(f"Тестовое изображение не найдено: {image_path}")
        return
    
    # UNet should output 1 channel for segmentation
    print("Инициализация моделей...")
    
 1
    # Let's modify the UNet to output 1 channel
    class FixedSimpleUNet(nn.Module):
        def __init__(self, in_channels=3, out_channels=1):
            super(FixedSimpleUNet, self).__init__()
            
            # Encoder
            self.enc1 = nn.Sequential(
                nn.Conv2d(in_channels, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(),
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU()
            )
            self.pool1 = nn.MaxPool2d(2)
            
            self.enc2 = nn.Sequential(
                nn.Conv2d(16, 32, 3, padding=1),
                nn.BatchNorm2d(32),
                nn.ReLU(),
                nn.Conv2d(32, 32, 3, padding=1),
                nn.BatchNorm2d(32),
                nn.ReLU()
            )
            self.pool2 = nn.MaxPool2d(2)
            
            # Bottleneck
            self.bottleneck = nn.Sequential(
                nn.Conv2d(32, 64, 3, padding=1),
                nn.BatchNorm2d(64),
                nn.ReLU(),
                nn.Conv2d(64, 64, 3, padding=1),
                nn.BatchNorm2d(64),
                nn.ReLU()
            )
            
            # Decoder
            self.up2 = nn.ConvTranspose2d(64, 32, 2, stride=2)
            self.dec2 = nn.Sequential(
                nn.Conv2d(64, 32, 3, padding=1),
                nn.BatchNorm2d(32),
                nn.ReLU(),
                nn.Conv2d(32, 32, 3, padding=1),
                nn.BatchNorm2d(32),
                nn.ReLU()
            )
            
            self.up1 = nn.ConvTranspose2d(32, 16, 2, stride=2)
            self.dec1 = nn.Sequential(
                nn.Conv2d(32, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(),
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU()
            )
            

            self.out = nn.Conv2d(16, out_channels, 1)
            
        def forward(self, x):
            # Encoder
            e1 = self.enc1(x)
            p1 = self.pool1(e1)
            
            e2 = self.enc2(p1)
            p2 = self.pool2(e2)
            
            # Bottleneck
            b = self.bottleneck(p2)
            
            # Decoder
            u2 = self.up2(b)
            if u2.shape != e2.shape:
                u2 = torch.nn.functional.interpolate(u2, size=e2.shape[2:], mode='bilinear')
            d2 = self.dec2(torch.cat([u2, e2], dim=1))
            
            u1 = self.up1(d2)
            if u1.shape != e1.shape:
                u1 = torch.nn.functional.interpolate(u1, size=e1.shape[2:], mode='bilinear')
            d1 = self.dec1(torch.cat([u1, e1], dim=1))
            
            return self.out(d1)
    
    unet = FixedSimpleUNet(in_channels=3, out_channels=1).to(device)
    generator = Generator(latent_dim=100, num_classes=5, img_channels=3).to(device)
    discriminator = Discriminator(num_classes=5, img_channels=3).to(device)
    

    print("Загрузка весов GAN...")
    checkpoint = torch.load(gan_model_path, map_location=device)
    
    
    generator_keys = ['generator', 'G', 'generator_state_dict']
    discriminator_keys = ['discriminator', 'D', 'discriminator_state_dict']
    
    
    generator_loaded = False
    for key in generator_keys:
        if key in checkpoint:
            try:
                generator.load_state_dict(checkpoint[key])
                print(f"Generator загружен из ключа: {key}")
                generator_loaded = True
                break
            except:
                continue
    
    if not generator_loaded:
        
        try:
            generator.load_state_dict(checkpoint)
            print("Generator загружен напрямую")
            generator_loaded = True
        except:
            print("Не удалось загрузить generator")

    
    discriminator_loaded = False
    for key in discriminator_keys:
        if key in checkpoint:
            try:
                discriminator.load_state_dict(checkpoint[key])
                print(f"Discriminator загружен из ключа: {key}")
                discriminator_loaded = True
                break
            except:
                continue
    
    if not discriminator_loaded:
        print("Не удалось загрузить discriminator")
    
    # Загрузка UNet если есть
    if os.path.exists(unet_model_path):
        print("Загрузка весов UNet...")
        # Load with strict=False to handle channel mismatch
        unet_state_dict = torch.load(unet_model_path, map_location=device)
        
        # Check if we need to modify the output layer weights
        if 'out.weight' in unet_state_dict:
            # Original has 3 output channels, we need 1
            if unet_state_dict['out.weight'].shape[0] == 3:
                # Take only the first channel
                unet_state_dict['out.weight'] = unet_state_dict['out.weight'][:1]
                unet_state_dict['out.bias'] = unet_state_dict['out.bias'][:1]
        
        unet.load_state_dict(unet_state_dict, strict=False)
        print("UNet weights loaded (with adjustments)")
    else:
        print("Модель UNet не найдена, будет использована случайно инициализированная")
    
    # Режим оценки
    generator.eval()
    discriminator.eval()
    unet.eval()
    
    print("Все модели успешно загружены!")
    
    
    transform = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
    ])
    
    
    print(f"Обработка изображения: {image_path}")
    image = Image.open(image_path).convert('RGB')
    original_size = image.size  # (width, height)
    
    image_tensor = transform(image).unsqueeze(0).to(device)
    
    
    anomaly_colors = {
        0: [0, 255, 0],    # Green - normal
        1: [255, 0, 0],    # Red
        2: [0, 0, 255],    # Blue  
        3: [255, 255, 0],  # Yellow
        4: [255, 0, 255],  # Purple
    }
    
    class_names = ['normal', 'abrasive', 'adhasive', 'cracks', 'delamination']
    
    
    with torch.no_grad():
        
        validity, class_probs = discriminator(image_tensor)
        predicted_class = torch.argmax(class_probs, dim=1).item()
        confidence = class_probs[0, predicted_class].item()
        
        
        z = torch.randn(1, 100).to(device)
        label_tensor = torch.tensor([predicted_class], device=device)
        reconstructed = generator(z, label_tensor)
        
        gan_difference = torch.abs(image_tensor - reconstructed)

        # UNet should output 1 channel now
        unet_mask = unet(image_tensor)
        
        # Debug shapes
        print(f"Input shape: {image_tensor.shape}")
        print(f"UNet mask shape: {unet_mask.shape}")
        print(f"GAN difference shape: {gan_difference.shape}")
        
        # Make sure UNet output is 1 channel
        if unet_mask.shape[1] != 1:
            unet_mask = unet_mask[:, :1]  # Take only first channel
        
        # GAN difference has 3 channels, take mean to get 1 channel
        gan_diff_single_channel = gan_difference.mean(dim=1, keepdim=True)
        
        # Now both have 1 channel, combine them
        combined_anomaly = (unet_mask + gan_diff_single_channel) / 2
        
        anomaly_mask = (combined_anomaly > 0.5).float()
        
        anomaly_mask_np = anomaly_mask.squeeze().cpu().numpy()
        
        # Resize mask to original image size
        anomaly_mask_resized = cv2.resize(anomaly_mask_np, original_size)  # original_size is (width, height)
        
        # GAN difference для визуализации
        gan_diff_visual = gan_difference.squeeze().mean(dim=0).cpu().numpy()
        unet_mask_visual = unet_mask.squeeze().cpu().numpy()
        combined_visual = combined_anomaly.squeeze().cpu().numpy()
    
    
    original_image = cv2.imread(image_path)
    original_image_rgb = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
    
    
    print(f"Original image shape: {original_image_rgb.shape}")
    print(f"Anomaly mask shape: {anomaly_mask_resized.shape}")
    
    
    color_mask = np.zeros_like(original_image_rgb)
    
    
    # Get boolean mask
    mask_bool = anomaly_mask_resized > 0
    
    # Apply color using numpy indexing
    for c in range(3):
        color_mask[:, :, c][mask_bool] = anomaly_colors[predicted_class][c]
    
    # Наложение
    result_image = cv2.addWeighted(original_image_rgb, 0.7, color_mask, 0.3, 0)
    
    
    fig, axes = plt.subplots(2, 4, figsize=(20, 10))
    axes = axes.flatten()
    
    # 1. 
    axes[0].imshow(original_image_rgb)
    axes[0].set_title('Original Image')
    axes[0].axis('off')
    
    # 2. 
    im2 = axes[1].imshow(gan_diff_visual, cmap='hot')
    axes[1].set_title('GAN Difference Map')
    plt.colorbar(im2, ax=axes[1])
    axes[1].axis('off')
    
    # 3. 
    im3 = axes[2].imshow(unet_mask_visual, cmap='hot')
    axes[2].set_title('UNet Mask')
    plt.colorbar(im3, ax=axes[2])
    axes[2].axis('off')
    
    # 4. 
    im4 = axes[3].imshow(combined_visual, cmap='hot')
    axes[3].set_title('Combined Anomaly Map')
    plt.colorbar(im4, ax=axes[3])
    axes[3].axis('off')
    
    # 5. 
    axes[4].imshow(anomaly_mask_resized, cmap='gray')
    axes[4].set_title('Binary Anomaly Mask')
    axes[4].axis('off')
    
    # 6. 
    axes[5].imshow(color_mask)
    axes[5].set_title('Color Mask')
    axes[5].axis('off')
    
    # 7. 
    axes[6].imshow(result_image)
    axes[6].set_title(f'Result\nClass: {class_names[predicted_class]}\nConfidence: {confidence:.3f}')
    axes[6].axis('off')
    
    # 8. 
    classes = list(range(len(class_names)))
    probs = class_probs[0].cpu().numpy()
    colors = ['green' if i == predicted_class else 'gray' for i in classes]
    axes[7].bar(classes, probs, color=colors)
    axes[7].set_xticks(classes)
    axes[7].set_xticklabels(class_names, rotation=45)
    axes[7].set_title('Class Probabilities')
    axes[7].set_ylim([0, 1])
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nПредсказанный класс: {class_names[predicted_class]}")
    print(f"Уверенность: {confidence:.3f}")
    print(f"Размер маски аномалий: {anomaly_mask_resized.shape}")
    
    print("\nВероятности по классам:")
    for i, class_name in enumerate(class_names):
        prob = class_probs[0, i].item()
        print(f"  {class_name}: {prob:.4f}")

if __name__ == "__main__":
    main()
```

```python
class SimpleUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super(SimpleUNet, self).__init__()
        
        # Encoder
        self.enc1 = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        self.pool1 = nn.MaxPool2d(2)
        
        self.enc2 = nn.Sequential(
            nn.Conv2d(16, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.pool2 = nn.MaxPool2d(2)
        
        # Bottleneck
        self.bottleneck = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU()
        )
        
        # Decoder
        self.up2 = nn.ConvTranspose2d(64, 32, 2, stride=2)
        self.dec2 = nn.Sequential(
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        
        self.up1 = nn.ConvTranspose2d(32, 16, 2, stride=2)
        self.dec1 = nn.Sequential(
            nn.Conv2d(32, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.Conv2d(16, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU()
        )
        
        # Output
        self.out = nn.Conv2d(16, 3, 1)
        
    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        p1 = self.pool1(e1)
        
        e2 = self.enc2(p1)
        p2 = self.pool2(e2)
        
        # Bottleneck
        b = self.bottleneck(p2)
        
        # Decoder
        u2 = self.up2(b)
        # Resize if needed
        if u2.shape != e2.shape:
            u2 = torch.nn.functional.interpolate(u2, size=e2.shape[2:], mode='bilinear')
        d2 = self.dec2(torch.cat([u2, e2], dim=1))
        
        u1 = self.up1(d2)
        if u1.shape != e1.shape:
            u1 = torch.nn.functional.interpolate(u1, size=e1.shape[2:], mode='bilinear')
        d1 = self.dec1(torch.cat([u1, e1], dim=1))
        
        return self.out(d1)


```

```python
class Generator(nn.Module):
    def __init__(self, latent_dim=100, num_classes=5, img_channels=3):
        super(Generator, self).__init__()
        self.num_classes = num_classes
        self.latent_dim = latent_dim
        
        self.init_size = 16
        self.label_embedding = nn.Embedding(num_classes, latent_dim)
        
        self.l1 = nn.Sequential(nn.Linear(latent_dim * 2, 128 * self.init_size ** 2))
        
        self.conv_blocks = nn.Sequential(
            nn.BatchNorm2d(128),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 128, 3, stride=1, padding=1),
            nn.BatchNorm2d(128, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout2d(0.1),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 64, 3, stride=1, padding=1),
            nn.BatchNorm2d(64, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, img_channels, 3, stride=1, padding=1),
            nn.Tanh()
        )
    
    def forward(self, z, labels):
        label_embed = self.label_embedding(labels)
        gen_input = torch.cat((label_embed, z), -1)
        
        out = self.l1(gen_input)
        out = out.view(out.shape[0], 128, self.init_size, self.init_size)
        img = self.conv_blocks(out)
        return img

class Discriminator(nn.Module):
    def __init__(self, num_classes=5, img_channels=3):
        super(Discriminator, self).__init__()
        
        def discriminator_block(in_filters, out_filters, bn=True):
            block = [nn.Conv2d(in_filters, out_filters, 3, 2, 1),
                    nn.LeakyReLU(0.2, inplace=True),
                    nn.Dropout2d(0.3)]
            if bn:
                block.append(nn.BatchNorm2d(out_filters, 0.8))
            return block
        
        self.model = nn.Sequential(
            *discriminator_block(img_channels, 16, bn=False),
            *discriminator_block(16, 32),
            *discriminator_block(32, 64),
            *discriminator_block(64, 128),
        )
        
        ds_size = 64 // 2 ** 4
        self.adv_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, 1), nn.Sigmoid())
        self.aux_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, num_classes), nn.Softmax(dim=1))
    
    def forward(self, img):
        out = self.model(img)
        out = out.view(out.shape[0], -1)
        validity = self.adv_layer(out)
        label = self.aux_layer(out)
        return validity, label


```

```python
class DefectDetector:
    def __init__(self, model_path, unet_path=None, img_size=64):
        self.img_size = img_size
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        
        # 
        self.generator = Generator(100, 5, 3).to(self.device)
        self.discriminator = Discriminator(5, 3).to(self.device)
        

        self.unet = SimpleUNet(in_channels=3, out_channels=1).to(self.device)
        
   
        checkpoint = torch.load(model_path, map_location=self.device)
        
        # 
        generator_loaded = False
        discriminator_loaded = False
        
        generator_keys = ['generator', 'G', 'generator_state_dict']
        discriminator_keys = ['discriminator', 'D', 'discriminator_state_dict']
        
        # r
        for key in generator_keys:
            if key in checkpoint:
                try:
                    self.generator.load_state_dict(checkpoint[key])
                    print(f"Generator загружен из ключа: {key}")
                    generator_loaded = True
                    break
                except:
                    continue
        
        if not generator_loaded:
            try:
                self.generator.load_state_dict(checkpoint)
                print("Generator загружен напрямую")
                generator_loaded = True
            except:
                print("Не удалось загрузить generator")
        
        # 
        for key in discriminator_keys:
            if key in checkpoint:
                try:
                    self.discriminator.load_state_dict(checkpoint[key])
                    print(f"Discriminator загружен из ключа: {key}")
                    discriminator_loaded = True
                    break
                except:
                    continue
        
        if not discriminator_loaded:
            print("Не удалось загрузить discriminator")
        
        # 
        if unet_path:
            try:
                unet_checkpoint = torch.load(unet_path, map_location=self.device)
                
                
                if 'out.weight' in unet_checkpoint:
                    
                    if unet_checkpoint['out.weight'].shape[0] == 3:
                        unet_checkpoint['out.weight'] = unet_checkpoint['out.weight'][:1]
                        if 'out.bias' in unet_checkpoint:
                            unet_checkpoint['out.bias'] = unet_checkpoint['out.bias'][:1]
                
                self.unet.load_state_dict(unet_checkpoint, strict=False)
                print(f"UNet загружен из: {unet_path}")
                self.unet_loaded = True
            except Exception as e:
                print(f"Ошибка загрузки UNet из {unet_path}: {e}")
                self.unet_loaded = False
        else:
            print("UNet не указан, используется только GAN")
            self.unet_loaded = False
        
        self.generator.eval()
        self.discriminator.eval()
        self.unet.eval()
        
        # 
        self.transform = transforms.Compose([
            transforms.Resize((img_size, img_size)),
            transforms.ToTensor(),
            transforms.Normalize([0.5] * 3, [0.5] * 3)
        ])
        
        # 
        self.defect_names = {
            0: "normal(green)",
            1: "abrasive(red)",
            2: "adhesive(blue)",
            3: "crack(yellow)",
            4: "delamination(pink)"
        }
        
        # 
        self.defect_colors = {
            0: [0, 255, 0],    # Green - normal
            1: [255, 0, 0],    # Red
            2: [0, 0, 255],    # Blue  
            3: [255, 255, 0],  # Yellow
            4: [255, 0, 255]   # Pink/Purple
        }

    def preprocess_image(self, image_path):
        
        image = cv2.imread(image_path)
        if image is None:
            raise ValueError(f"Не удалось загрузить изображение: {image_path}")
        return cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    def process_tiles(self, image, tile_size=64, stride=32):
        
        h, w = image.shape[:2]
        tiles = []
        positions = []
        
        
        for y in range(0, h, stride):
            for x in range(0, w, stride):
                # 
                y_end = min(y + tile_size, h)
                x_end = min(x + tile_size, w)
                
                # 
                if (y_end - y) < tile_size // 2 or (x_end - x) < tile_size // 2:
                    continue
                
                tile = image[y:y_end, x:x_end]
                
                # 
                if tile.shape[0] != tile_size or tile.shape[1] != tile_size:
                    tile = cv2.resize(tile, (tile_size, tile_size))
                
                tiles.append(tile)
                positions.append((x, y, x_end, y_end))
        
        return tiles, positions

    def analyze_tile(self, tile_rgb):
        """Анализ одного тайла"""
        pil_image = Image.fromarray(tile_rgb)
        image_tensor = self.transform(pil_image).unsqueeze(0).to(self.device)
        
        with torch.no_grad():
            # 
            validity, class_probs = self.discriminator(image_tensor)
            predicted_class = torch.argmax(class_probs, dim=1).item()
            confidence = class_probs[0, predicted_class].item()
            
            # F
            if predicted_class != 0 and confidence > 0.7:  
                # 
                z = torch.randn(1, 100).to(self.device)
                label_tensor = torch.tensor([predicted_class], device=self.device)
                reconstructed = self.generator(z, label_tensor)
                
                # 
                gan_diff = torch.abs(image_tensor - reconstructed).mean(dim=1, keepdim=True)
                
                if self.unet_loaded:
                    try:
                        unet_mask = self.unet(image_tensor)
                        # 
                        if unet_mask.shape[1] != 1:
                            unet_mask = unet_mask[:, :1]
                        anomaly_map = (gan_diff + unet_mask) / 2
                    except:
                        anomaly_map = gan_diff
                else:
                    anomaly_map = gan_diff
                
                # 
                if anomaly_map.max() > 0:
                    anomaly_map = anomaly_map / anomaly_map.max()
                
                anomaly_np = anomaly_map.squeeze().cpu().numpy()
                
                return {
                    'class': predicted_class,
                    'confidence': confidence,
                    'anomaly_map': anomaly_np,
                    'is_defect': True
                }
            else:
                # 
                return {
                    'class': 0,  # 
                    'confidence': confidence,
                    'anomaly_map': None,
                    'is_defect': False
                }

    def detect_defects(self, image_path):
        
        original_rgb = self.preprocess_image(image_path)
        h, w = original_rgb.shape[:2]
        
        # 
        defect_mask_float = np.zeros((h, w, 3), dtype=np.float32)
        defect_counts = {0: 0, 1: 0, 2: 0, 3: 0, 4: 0}
        defect_confidence_sum = {0: 0, 1: 0, 2: 0, 3: 0, 4: 0}
        
        total_tiles = 0
        defect_tiles = 0
        
        tiles, positions = self.process_tiles(original_rgb, tile_size=self.img_size, stride=self.img_size//2)
        total_tiles = len(tiles)
        
        print(f"Обработка {total_tiles} тайлов...")
        
        for tile, (x1, y1, x2, y2) in zip(tiles, positions):
            result = self.analyze_tile(tile)
            
            if result is not None:
                class_id = result['class']
                defect_counts[class_id] += 1
                defect_confidence_sum[class_id] += result['confidence']
                
                if result['is_defect']:
                    defect_tiles += 1
                    
                    if result['anomaly_map'] is not None:
                        anomaly_map = result['anomaly_map']
                        tile_h, tile_w = anomaly_map.shape
                        
                        
                        if anomaly_map.max() > 0:
                            threshold = np.percentile(anomaly_map, 85)  
                            defect_pixels = anomaly_map > threshold
                            
                            if np.any(defect_pixels):
                                color = self.defect_colors[class_id]
                                
                                # 
                                actual_h = y2 - y1
                                actual_w = x2 - x1
                                
                                if actual_h != tile_h or actual_w != tile_w:
                                    defect_pixels_resized = cv2.resize(
                                        defect_pixels.astype(np.float32), 
                                        (actual_w, actual_h)
                                    ) > 0.5
                                else:
                                    defect_pixels_resized = defect_pixels
                                
                                # 
                                for c in range(3):
                                    defect_mask_float[y1:y2, x1:x2, c][defect_pixels_resized] += color[c]
        
        # 
        defect_mask = np.clip(defect_mask_float, 0, 255).astype(np.uint8)
        
        # 
        total_pixels = h * w
        defect_percentages = {}
        
        # 
        for class_id in range(1, 5):
            if class_id in self.defect_colors:
                color = np.array(self.defect_colors[class_id])
                # 
                if np.any(defect_mask > 0):
                    color_mask = np.all(np.abs(defect_mask - color) < 50, axis=2)
                    defect_counts[class_id] = np.sum(color_mask)
                    defect_percentages[class_id] = (defect_counts[class_id] / total_pixels) * 100
        
        # 
        if total_tiles > 0:
            defect_tile_percentage = (defect_tiles / total_tiles) * 100
            normal_tile_percentage = 100 - defect_tile_percentage
        else:
            defect_tile_percentage = 0
            normal_tile_percentage = 100
        
        # 
        total_defect_pixels = sum(defect_counts.get(i, 0) for i in range(1, 5))
        if total_pixels > 0:
            defect_pixel_percentage = (total_defect_pixels / total_pixels) * 100
            normal_pixel_percentage = 100 - defect_pixel_percentage
        else:
            defect_pixel_percentage = 0
            normal_pixel_percentage = 100
        
        return (original_rgb, defect_mask, defect_counts, defect_percentages, 
                normal_tile_percentage, defect_tile_percentage,
                normal_pixel_percentage, defect_pixel_percentage)

    def show_results(self, image_path):
        """ """
        (original_rgb, defect_mask, defect_counts, defect_percentages, 
         normal_tile_percent, defect_tile_percent,
         normal_pixel_percent, defect_pixel_percent) = self.detect_defects(image_path)
        
        # 
        result_image = cv2.addWeighted(original_rgb, 0.7, defect_mask, 0.3, 0)
        
    
        
        print(f"\nshape: {self.img_size}x{self.img_size}):")
        print(f"  Normal area: {normal_tile_percent:.1f}%")
        print(f"  Defects area: {defect_tile_percent:.1f}%")
    
        
        print("\nDefects detalization:")
        found_defects = False
        for class_id in [2, 1, 3, 4]:  # 
            if class_id in defect_percentages and defect_percentages[class_id] > 0.01:
                name = self.defect_names[class_id]
                percentage = defect_percentages[class_id]
                count = defect_counts.get(class_id, 0)
                print(f"  {name}: {percentage:.2f}% ({count} pixels)")
                found_defects = True

        
        # 
        fig, axes = plt.subplots(1, 4, figsize=(20, 5))
        
        axes[0].imshow(original_rgb)
        axes[0].set_title('Original image')
        axes[0].axis('off')
        
        axes[1].imshow(defect_mask)
        axes[1].set_title('Defects map')
        axes[1].axis('off')
        
        axes[2].imshow(result_image)
        axes[2].set_title('Result')
        axes[2].axis('off')
        
        # 
        if found_defects:
            defect_classes = []
            defect_values = []
            colors = []
            
            for class_id in [1, 2, 3, 4]:
                if class_id in defect_percentages and defect_percentages[class_id] > 0:
                    defect_classes.append(self.defect_names[class_id].split('(')[0])
                    defect_values.append(defect_percentages[class_id])
                    colors.append([c/255 for c in self.defect_colors[class_id]])
            
            if defect_values:
                axes[3].bar(defect_classes, defect_values, color=colors)
                axes[3].set_title('Defects(%)')
                axes[3].set_ylabel('Percentage of area')
                axes[3].tick_params(axis='x', rotation=45)
                axes[3].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        return result_image, defect_mask, normal_pixel_percent, defect_pixel_percent
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms
import os
from graphviz import Digraph
import sys

# Check if graphviz is installed
try:
    import graphviz
    GRAPHVIZ_AVAILABLE = True
except ImportError:
    print("Graphviz not installed. Installing...")
    try:
        import subprocess
        subprocess.check_call([sys.executable, "-m", "pip", "install", "graphviz"])
        import graphviz
        GRAPHVIZ_AVAILABLE = True
        print("Graphviz installed successfully!")
    except:
        GRAPHVIZ_AVAILABLE = False
        print("Could not install Graphviz. Please install manually: pip install graphviz")
        print("Also install Graphviz software from: https://graphviz.org/download/")

# Assuming you have these classes defined or will import them
class SimpleUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super().__init__()
        # Add your actual UNet architecture here
        self.encoder = nn.Sequential(
            nn.Conv2d(in_channels, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True)
        )
        self.decoder = nn.Sequential(
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, out_channels, 1)
        )
        
    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return torch.sigmoid(x)

class Generator(nn.Module):
    def __init__(self, latent_dim=100, num_classes=5, channels=3):
        super().__init__()
        self.init_size = 64 // 4
        self.l1 = nn.Sequential(
            nn.Linear(latent_dim + num_classes, 128 * self.init_size ** 2)
        )
        
        self.conv_blocks = nn.Sequential(
            nn.BatchNorm2d(128),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 128, 3, stride=1, padding=1),
            nn.BatchNorm2d(128, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Upsample(scale_factor=2),
            nn.Conv2d(128, 64, 3, stride=1, padding=1),
            nn.BatchNorm2d(64, 0.8),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, channels, 3, stride=1, padding=1),
            nn.Tanh()
        )
        
    def forward(self, z, labels):
        # Concatenate label embedding and noise
        gen_input = torch.cat((z, labels), -1)
        out = self.l1(gen_input)
        out = out.view(out.shape[0], 128, self.init_size, self.init_size)
        img = self.conv_blocks(out)
        return img

class Discriminator(nn.Module):
    def __init__(self, num_classes=5, channels=3):
        super().__init__()
        
        def discriminator_block(in_filters, out_filters, bn=True):
            block = [nn.Conv2d(in_filters, out_filters, 3, 2, 1),
                    nn.LeakyReLU(0.2, inplace=True),
                    nn.Dropout2d(0.25)]
            if bn:
                block.append(nn.BatchNorm2d(out_filters, 0.8))
            return block
        
        self.model = nn.Sequential(
            *discriminator_block(channels, 16, bn=False),
            *discriminator_block(16, 32),
            *discriminator_block(32, 64),
            *discriminator_block(64, 128)
        )
        
        # Calculate output shape
        ds_size = 64 // 2 ** 4
        self.adv_layer = nn.Sequential(
            nn.Linear(128 * ds_size ** 2, 1),
            nn.Sigmoid()
        )
        self.aux_layer = nn.Sequential(
            nn.Linear(128 * ds_size ** 2, num_classes),
            nn.Softmax(dim=1)
        )
        
    def forward(self, img):
        out = self.model(img)
        out = out.view(out.shape[0], -1)
        validity = self.adv_layer(out)
        label = self.aux_layer(out)
        return validity, label

class ACGANAnomalyDetectionVisualizer:
    def __init__(self, num_classes=5, img_size=64, channels=3, latent_dim=100,
                 model_path=None, unet_path=None):
            return
            
        self.num_classes = num_classes
        self.img_size = img_size
        self.channels = channels
        self.latent_dim = latent_dim
        
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"📊 ACGAN Model Visualizer")
        print(f"📁 Model path: {model_path}")
        print(f"📁 UNet path: {unet_path}")
        print(f"⚙️ Device: {self.device}")
        print("-" * 60)

        # Initialize models
        self.unet = SimpleUNet(in_channels=3, out_channels=1).to(self.device)
        self.generator = Generator(latent_dim, num_classes, channels).to(self.device)
        self.discriminator = Discriminator(num_classes, channels).to(self.device)
        
        # Load weights if paths provided
        self.load_weights(model_path, unet_path)
        
        # Print model info
        self.print_model_info()

    def load_weights(self, model_path, unet_path):
        """Load pretrained weights if available"""
        # Load UNet weights
        if unet_path and os.path.exists(unet_path):
            print(f"🔹 Loading UNet from: {unet_path}")
            try:
                self.unet.load_state_dict(torch.load(unet_path, map_location=self.device))
                print("   ✅ UNet loaded successfully")
            except Exception as e:
                print(f"   ❌ Failed to load UNet: {e}")
        else:
            print("   ⚠️ UNet path not provided or file doesn't exist")
        
        # Load full model weights
        if model_path and os.path.exists(model_path):
            print(f"🔹 Loading full model from: {model_path}")
            try:
                checkpoint = torch.load(model_path, map_location=self.device)
                
                # Try different loading strategies
                if 'unet_state_dict' in checkpoint:
                    self.unet.load_state_dict(checkpoint['unet_state_dict'])
                    print("   ✅ UNet loaded from checkpoint")
                if 'generator_state_dict' in checkpoint:
                    self.generator.load_state_dict(checkpoint['generator_state_dict'])
                    print("   ✅ Generator loaded from checkpoint")
                if 'discriminator_state_dict' in checkpoint:
                    self.discriminator.load_state_dict(checkpoint['discriminator_state_dict'])
                    print("   ✅ Discriminator loaded from checkpoint")
                elif 'state_dict' in checkpoint:
                    try:
                        # This would require the complete ACGANAnomalyDetection class
                        print("   ⚠️ Full state_dict found but cannot load without original class")
                    except:
                        print("   ⚠️ Could not load full state_dict")
                print("   ✅ Model weights loaded")
            except Exception as e:
                print(f"   ❌ Failed to load model: {e}")
        else:
            print("   ⚠️ Model path not provided or file doesn't exist")

    def print_model_info(self):
        """Print basic model information"""
        print("\n📈 MODEL INFORMATION:")
        print("-" * 40)
        
        def count_params(model):
            return sum(p.numel() for p in model.parameters())
        
        g_params = count_params(self.generator)
        d_params = count_params(self.discriminator)
        u_params = count_params(self.unet)
        total_params = g_params + d_params + u_params
        
        print(f"Generator parameters: {g_params:,}")
        print(f"Discriminator parameters: {d_params:,}")
        print(f"UNet parameters: {u_params:,}")
        print(f"Total parameters: {total_params:,}")
        print(f"Image size: {self.img_size}x{self.img_size}")
        print(f"Latent dimension: {self.latent_dim}")
        print(f"Number of classes: {self.num_classes}")

    def visualize_combined_structure(self, output_path="combined_model_structure"):
        """Visualize the complete combined model structure"""
        if not GRAPHVIZ_AVAILABLE:
            return
            
        print(f"\n🎨 Creating combined model visualization...")
        
        dot = Digraph(comment='ACGAN Anomaly Detection Combined Model', format='png')
        dot.attr(rankdir='TB', size='20,25', label='ACGAN ANOMALY DETECTION\nCombined Model Structure', labelloc='t', fontsize='24')
        
        # Main model container
        with dot.subgraph(name='cluster_main') as c:
            c.attr(label='ACGAN Anomaly Detection System', style='filled', color='lightblue', fontsize='20')
            
            # Input section
            with c.subgraph(name='cluster_inputs') as input_c:
                input_c.attr(label='Inputs', style='filled', color='lightgreen')
                input_c.node('noise', 'Noise Vector\n(100-dim)', shape='ellipse')
                input_c.node('label', 'Class Label\n(one-hot)', shape='ellipse')
                input_c.node('image', 'Input Image\n(3, 64, 64)', shape='cylinder')
            
            # Generator section
            with c.subgraph(name='cluster_generator') as gen_c:
                gen_c.attr(label='Generator (Conditional GAN)', style='filled', color='lightyellow')
                gen_c.node('G_input', 'Concatenate\n[z + label]', shape='box')
                gen_c.node('G_fc', 'Linear Layer\n128 × (16×16)', shape='box')
                gen_c.node('G_reshape', 'Reshape\n(128, 16, 16)', shape='box')
                gen_c.node('G_conv1', 'Upsample + Conv\nBatchNorm + LeakyReLU', shape='box')
                gen_c.node('G_conv2', 'Upsample + Conv\nBatchNorm + LeakyReLU', shape='box')
                gen_c.node('G_output', 'Conv + Tanh\nGenerated Image', shape='box')
                
                # Connect generator
                gen_c.edge('G_input', 'G_fc')
                gen_c.edge('G_fc', 'G_reshape')
                gen_c.edge('G_reshape', 'G_conv1')
                gen_c.edge('G_conv1', 'G_conv2')
                gen_c.edge('G_conv2', 'G_output')
            
            # Discriminator section
            with c.subgraph(name='cluster_discriminator') as disc_c:
                disc_c.attr(label='Discriminator (Classifier)', style='filled', color='lightpink')
                disc_c.node('D_input', 'Conv Block 1\n16 filters', shape='box')
                disc_c.node('D_conv2', 'Conv Block 2\n32 filters', shape='box')
                disc_c.node('D_conv3', 'Conv Block 3\n64 filters', shape='box')
                disc_c.node('D_conv4', 'Conv Block 4\n128 filters', shape='box')
                disc_c.node('D_flatten', 'Flatten', shape='box')
                disc_c.node('D_adv', 'Linear + Sigmoid\nReal/Fake', shape='box')
                disc_c.node('D_aux', 'Linear + Softmax\nClass (5)', shape='box')
                
                # Connect discriminator
                disc_c.edge('D_input', 'D_conv2')
                disc_c.edge('D_conv2', 'D_conv3')
                disc_c.edge('D_conv3', 'D_conv4')
                disc_c.edge('D_conv4', 'D_flatten')
                disc_c.edge('D_flatten', 'D_adv')
                disc_c.edge('D_flatten', 'D_aux')
            
            # UNet section
            with c.subgraph(name='cluster_unet') as unet_c:
                unet_c.attr(label='UNet (Segmentation)', style='filled', color='lightgrey')
                unet_c.node('U_encoder', 'Encoder\nConv + ReLU ×2', shape='box')
                unet_c.node('U_decoder', 'Decoder\nConv + ReLU', shape='box')
                unet_c.node('U_output', 'Conv + Sigmoid\nAnomaly Mask', shape='box')
                
                # Connect UNet
                unet_c.edge('U_encoder', 'U_decoder')
                unet_c.edge('U_decoder', 'U_output')
            
            # Output section
            with c.subgraph(name='cluster_outputs') as out_c:
                out_c.attr(label='Outputs & Processing', style='filled', color='lightcoral')
                out_c.node('gen_img', 'Generated Image', shape='cylinder')
                out_c.node('difference', 'GAN Difference\n|Real - Generated|', shape='diamond')
                out_c.node('unet_mask', 'UNet Mask', shape='parallelogram')
                out_c.node('combine', 'Combine Masks\n(UNet + GAN)', shape='diamond')
                out_c.node('threshold', 'Threshold\n(0.5)', shape='box')
                out_c.node('final_mask', 'Final Anomaly Mask', shape='parallelogram')
                out_c.node('class_pred', 'Class Prediction', shape='note')
                
                # Connect outputs
                out_c.edge('gen_img', 'difference')
                out_c.edge('unet_mask', 'combine')
                out_c.edge('difference', 'combine')
                out_c.edge('combine', 'threshold')
                out_c.edge('threshold', 'final_mask')
        
        # Connect main flow
        dot.edge('noise', 'G_input')
        dot.edge('label', 'G_input')
        dot.edge('G_output', 'gen_img')
        dot.edge('image', 'D_input')
        dot.edge('image', 'U_encoder')
        dot.edge('image', 'difference')
        dot.edge('D_aux', 'class_pred')
        dot.edge('U_output', 'unet_mask')
        
        # Save and render
        try:
            dot.render(output_path, view=True, cleanup=False)
            print(f"✅ Combined model visualization saved as: {output_path}.png")
            return dot
        except Exception as e:
            print(f"❌ Failed to create visualization: {e}")
            return None

    def visualize_training_flow(self, output_path="training_flow"):
        """Visualize the training flow"""
        if not GRAPHVIZ_AVAILABLE:
            return
            
        print(f"\n🎨 Creating training flow visualization...")
        
        dot = Digraph(comment='ACGAN Training Flow', format='png')
        dot.attr(rankdir='TB', size='25,30', label='ACGAN TRAINING FLOW', labelloc='t', fontsize='24')
        
        # Training steps
        steps = [
            ('init', 'Initialize Models\n(G, D, UNet)', 'oval'),
            ('load_data', 'Load Training Data\n(Images + Labels)', 'cylinder'),
            ('batch', 'Create Batch\n(batch_size=32)', 'box'),
            ('generate', 'Generate Fake Images\nG(z, labels)', 'box'),
            ('unet_fwd', 'UNet Forward Pass\nPredict Masks', 'box'),
            ('pseudo_gt', 'Create Pseudo-GT\nfrom GAN difference', 'diamond'),
            ('loss_g', 'Calculate G Loss\n(Adv + Aux)', 'diamond'),
            ('loss_u', 'Calculate UNet Loss\n(BCEWithLogits)', 'diamond'),
            ('backward_gu', 'Backward Pass\n(G + UNet)', 'box'),
            ('optim_gu', 'Optimizer Step\n(G + UNet)', 'box'),
            ('disc_fwd', 'Discriminator Forward\n(Real + Fake)', 'box'),
            ('loss_d', 'Calculate D Loss\n(Real/Fake + Class)', 'diamond'),
            ('backward_d', 'Backward Pass (D)', 'box'),
            ('optim_d', 'Optimizer Step (D)', 'box'),
            ('update', 'Update Parameters', 'box'),
            ('next_batch', 'Next Batch', 'oval'),
            ('epoch_end', 'Epoch Complete\nSave Checkpoint', 'parallelogram')
        ]
        
        # Add nodes
        for name, label, shape in steps:
            dot.node(name, label, shape=shape)
        
        # Connect steps
        connections = [
            ('init', 'load_data'),
            ('load_data', 'batch'),
            ('batch', 'generate'),
            ('batch', 'unet_fwd'),
            ('generate', 'pseudo_gt'),
            ('unet_fwd', 'loss_u'),
            ('pseudo_gt', 'loss_u'),
            ('generate', 'loss_g'),
            ('loss_g', 'backward_gu'),
            ('loss_u', 'backward_gu'),
            ('backward_gu', 'optim_gu'),
            ('batch', 'disc_fwd'),
            ('generate', 'disc_fwd'),
            ('disc_fwd', 'loss_d'),
            ('loss_d', 'backward_d'),
            ('backward_d', 'optim_d'),
            ('optim_gu', 'update'),
            ('optim_d', 'update'),
            ('update', 'next_batch'),
            ('next_batch', 'batch'),
            ('next_batch', 'epoch_end')
        ]
        
        for src, dst in connections:
            dot.edge(src, dst)
        
        # Add annotations
        with dot.subgraph(name='cluster_losses') as c:
            c.attr(label='Loss Functions', style='dashed')
            c.node('L_adv', 'Adversarial Loss\n(BCELoss)', shape='note')
            c.node('L_aux', 'Auxiliary Loss\n(CrossEntropy)', shape='note')
            c.node('L_unet', 'UNet Loss\n(BCEWithLogits)', shape='note')
        
        # Save and render
        try:
            dot.render(output_path, view=True, cleanup=False)
            print(f"✅ Training flow visualization saved as: {output_path}.png")
            return dot
        except Exception as e:
            print(f"❌ Failed to create visualization: {e}")
            return None

    def visualize_component_details(self, output_path="component_details"):
        """Visualize detailed component architectures"""
        if not GRAPHVIZ_AVAILABLE:
            return
            
        print(f"\n🎨 Creating component details visualization...")
        
        dot = Digraph(comment='Model Component Details', format='png')
        dot.attr(rankdir='LR', size='30,20', label='MODEL COMPONENT DETAILS', labelloc='t', fontsize='24')
        
        # Generator details
        with dot.subgraph(name='cluster_generator') as c:
            c.attr(label='Generator Architecture', style='filled', color='lightyellow')
            c.node('G1', 'Input:\nz(100) + label(5)', shape='ellipse')
            c.node('G2', 'Linear Layer\n100+5 → 128×256', shape='box')
            c.node('G3', 'Reshape\n(128, 16, 16)', shape='box')
            c.node('G4', 'Upsample 2×\nConv2d 128→128', shape='box')
            c.node('G5', 'BatchNorm + LeakyReLU', shape='box')
            c.node('G6', 'Upsample 2×\nConv2d 128→64', shape='box')
            c.node('G7', 'BatchNorm + LeakyReLU', shape='box')
            c.node('G8', 'Conv2d 64→3\nTanh Activation', shape='box')
            c.node('G9', 'Output: Generated Image\n(3, 64, 64)', shape='cylinder')
            
            # Connect
            c.edges([('G1', 'G2'), ('G2', 'G3'), ('G3', 'G4'), ('G4', 'G5'),
                    ('G5', 'G6'), ('G6', 'G7'), ('G7', 'G8'), ('G8', 'G9')])
        
        # Discriminator details
        with dot.subgraph(name='cluster_discriminator') as c:
            c.attr(label='Discriminator Architecture', style='filled', color='lightpink')
            c.node('D1', 'Input Image\n(3, 64, 64)', shape='cylinder')
            c.node('D2', 'Conv2d 3→16\nstride=2, LeakyReLU', shape='box')
            c.node('D3', 'Dropout2d 0.25', shape='box')
            c.node('D4', 'Conv2d 16→32\nstride=2, BatchNorm', shape='box')
            c.node('D5', 'LeakyReLU + Dropout', shape='box')
            c.node('D6', 'Conv2d 32→64\nstride=2, BatchNorm', shape='box')
            c.node('D7', 'LeakyReLU + Dropout', shape='box')
            c.node('D8', 'Conv2d 64→128\nstride=2, BatchNorm', shape='box')
            c.node('D9', 'LeakyReLU + Dropout', shape='box')
            c.node('D10', 'Flatten', shape='box')
            c.node('D11', 'Linear + Sigmoid\nReal/Fake Score', shape='box')
            c.node('D12', 'Linear + Softmax\nClass Probabilities', shape='box')
            c.node('D13', 'Outputs:\nValidity + Class', shape='parallelogram')
            
            # Connect
            c.edges([('D1', 'D2'), ('D2', 'D3'), ('D3', 'D4'), ('D4', 'D5'),
                    ('D5', 'D6'), ('D6', 'D7'), ('D7', 'D8'), ('D8', 'D9'),
                    ('D9', 'D10'), ('D10', 'D11'), ('D10', 'D12'),
                    ('D11', 'D13'), ('D12', 'D13')])
        
        # UNet details
        with dot.subgraph(name='cluster_unet') as c:
            c.attr(label='UNet Architecture', style='filled', color='lightgrey')
            c.node('U1', 'Input Image\n(3, 64, 64)', shape='cylinder')
            c.node('U2', 'Encoder:\nConv2d 3→64', shape='box')
            c.node('U3', 'ReLU Activation', shape='box')
            c.node('U4', 'Conv2d 64→64', shape='box')
            c.node('U5', 'ReLU Activation', shape='box')
            c.node('U6', 'Decoder:\nConv2d 64→64', shape='box')
            c.node('U7', 'ReLU Activation', shape='box')
            c.node('U8', 'Conv2d 64→1', shape='box')
            c.node('U9', 'Sigmoid Activation', shape='box')
            c.node('U10', 'Output: Anomaly Mask\n(1, 64, 64)', shape='parallelogram')
            
            # Connect
            c.edges([('U1', 'U2'), ('U2', 'U3'), ('U3', 'U4'), ('U4', 'U5'),
                    ('U5', 'U6'), ('U6', 'U7'), ('U7', 'U8'), ('U8', 'U9'),
                    ('U9', 'U10')])
        
        # Save and render
        try:
            dot.render(output_path, view=True, cleanup=False)
            print(f"✅ Component details visualization saved as: {output_path}.png")
            return dot
        except Exception as e:
            print(f"❌ Failed to create visualization: {e}")
            return None

    def create_all_visualizations(self, output_dir="model_visualizations"):
        """Create all visualizations"""
        if not GRAPHVIZ_AVAILABLE:
            return
            
        # Create output directory
        os.makedirs(output_dir, exist_ok=True)
        
        print(f"\n📁 Creating all visualizations in: {output_dir}")
        
        # Create all visualizations
        self.visualize_combined_structure(os.path.join(output_dir, "combined_structure"))
        self.visualize_training_flow(os.path.join(output_dir, "training_flow"))
        self.visualize_component_details(os.path.join(output_dir, "component_details"))
        
        print(f"\n✅ All visualizations created in: {output_dir}")

# Usage example
if __name__ == "__main__":
    # Update these paths to your actual model files
    MODEL_PATH = r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\final_hybrid_model.pth"
    UNET_PATH = r"C:\Users\Administrator\OneDrive\Desktop\diana\Gan Anomaly\2_unet_trained.pth"
    
    # Create visualizer
    visualizer = ACGANAnomalyDetectionVisualizer(
        num_classes=5,
        img_size=64,
        channels=3,
        latent_dim=100,
        model_path=MODEL_PATH,
        unet_path=UNET_PATH
    )
    
    # Create visualizations
    visualizer.create_all_visualizations()
    
    print("\n" + "="*60)
    print("VISUALIZATION COMPLETE!")
    print("="*60)
    print("\nFiles created:")
    print("1. combined_structure.png - Complete system architecture")
    print("2. training_flow.png - Training process step-by-step")
    print("3. component_details.png - Detailed component architectures")
    print("\nOpen these files to see your model structure!")
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
from torchvision import transforms
from PIL import Image
import os
import glob
        
# Your discriminator architecture
class Discriminator(nn.Module):
    def __init__(self, num_classes=5, img_channels=3):
        super(Discriminator, self).__init__()
        
        def discriminator_block(in_filters, out_filters, bn=True):
            block = [nn.Conv2d(in_filters, out_filters, 3, 2, 1),
                    nn.LeakyReLU(0.2, inplace=True),
                    nn.Dropout2d(0.3)]
            if bn:
                block.append(nn.BatchNorm2d(out_filters, 0.8))
            return block
        
        self.model = nn.Sequential(
            *discriminator_block(img_channels, 16, bn=False),
            *discriminator_block(16, 32),
            *discriminator_block(32, 64),
            *discriminator_block(64, 128),
        )
        
        ds_size = 64 // 2 ** 4
        self.adv_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, 1), nn.Sigmoid())
        self.aux_layer = nn.Sequential(nn.Linear(128 * ds_size ** 2, num_classes), nn.Softmax(dim=1))
    
    def forward(self, img):
        out = self.model(img)
        out = out.view(out.shape[0], -1)
        validity = self.adv_layer(out)
        label = self.aux_layer(out)
        return validity, label

def load_discriminator(model_path):
    """Load discriminator for feature extraction"""
    discriminator = Discriminator(num_classes=5)
    checkpoint = torch.load(model_path, map_location='cpu')
    discriminator.load_state_dict(checkpoint['discriminator'])
    discriminator.eval()
    return discriminator

def extract_features_from_images(discriminator, folder_paths, class_names, num_images_per_class=20):
    """Extract features from real images using the discriminator's intermediate layers"""
    transform = transforms.Compose([
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize([0.5] * 3, [0.5] * 3)
    ])
    
    all_features = []
    all_labels = []
    all_filenames = []
    
    print("Extracting features from images...")
    
    for class_idx, folder_path in enumerate(folder_paths):
        print(f"Processing {class_names[class_idx]}...")
        
        # Check if folder exists
        if not os.path.exists(folder_path):
            print(f"Warning: Folder {folder_path} does not exist!")
            continue
        
        # Get images from folder
        extensions = ['*.jpg', '*.jpeg', '*.png', '*.bmp']
        images = []
        for ext in extensions:
            images.extend(glob.glob(os.path.join(folder_path, ext)))
        
        print(f"Found {len(images)} images in {class_names[class_idx]}")
        
        # Use available images (up to num_images_per_class)
        images = images[:num_images_per_class]
        
        for img_path in images:
            try:
                # Load and preprocess image
                img = Image.open(img_path).convert('RGB')
                img_tensor = transform(img).unsqueeze(0)
                
                # Extract features by running only the convolutional part
                with torch.no_grad():
                    # Get features from the convolutional layers
                    conv_features = discriminator.model(img_tensor)
                    # Flatten the features
                    features = conv_features.view(conv_features.size(0), -1)
                
                all_features.append(features.squeeze().numpy())
                all_labels.append(class_idx)
                all_filenames.append(os.path.basename(img_path))
                
            except Exception as e:
                print(f"Error processing {img_path}: {e}")
    
    # Check if we extracted any features
    if len(all_features) == 0:
        print("No features were extracted! Check your image paths.")
        return np.array([]), np.array([]), []
    
    return np.array(all_features), np.array(all_labels), all_filenames

def visualize_feature_space(features, labels, filenames, class_names, method='tsne'):
    """Visualize the feature space using dimensionality reduction"""
    
    print(f"Reducing dimensions using {method.upper()}...")
    
    if method == 'tsne':
        reducer = TSNE(n_components=2, random_state=42, perplexity=min(30, len(features)-1))
        embeddings = reducer.fit_transform(features)
        title = 't-SNE Visualization of Feature Space'
    elif method == 'pca':
        reducer = PCA(n_components=2, random_state=42)
        embeddings = reducer.fit_transform(features)
        variance_ratio = reducer.explained_variance_ratio_
        title = f'PCA Visualization (Variance: {variance_ratio[0]:.2f}, {variance_ratio[1]:.2f})'
    else:
        raise ValueError("Method must be 'tsne' or 'pca'")
    
    # Create visualization
    plt.figure(figsize=(12, 10))
    
    # Define colors for each class
    colors = ['blue', 'red', 'green', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan']
    
    for class_idx in range(len(class_names)):
        mask = labels == class_idx
        plt.scatter(embeddings[mask, 0], embeddings[mask, 1], 
                   c=colors[class_idx % len(colors)], 
                   label=class_names[class_idx],
                   alpha=0.7, s=50, edgecolors='black', linewidth=0.5)
    
    plt.title(title, fontsize=16, fontweight='bold')
    plt.xlabel('Component 1')
    plt.ylabel('Component 2')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Add some statistics
    unique, counts = np.unique(labels, return_counts=True)
    plt.figtext(0.02, 0.02, f"Total samples: {len(features)}\n" +
               "\n".join([f"{class_names[idx]}: {count}" for idx, count in zip(unique, counts)]),
               fontsize=20, bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
    
    plt.tight_layout()
    plt.show()
    
    return embeddings

def analyze_clusters(embeddings, labels, class_names):
    """Analyze the clustering behavior"""
    from sklearn.metrics import silhouette_score, calinski_harabasz_score
    
    print("\n=== CLUSTER ANALYSIS ===")
    
    # Calculate clustering metrics
    if len(np.unique(labels)) > 1:
        silhouette = silhouette_score(embeddings, labels)
        calinski = calinski_harabasz_score(embeddings, labels)
        
        print(f"Silhouette Score: {silhouette:.3f}")
        print(f"Calinski-Harabasz Score: {calinski:.3f}")
        
        # Interpretation
        if silhouette > 0.5:
            print("✓ Good clustering structure")
        elif silhouette > 0.25:
            print("○ Fair clustering structure") 
        else:
            print("✗ Weak clustering structure")
    
    # Calculate class separation
    print("\n=== CLASS SEPARATION ===")
    from scipy.spatial.distance import cdist
    
    # Calculate centroids
    centroids = []
    for class_idx in range(len(class_names)):
        mask = labels == class_idx
        if np.sum(mask) > 0:
            centroid = np.mean(embeddings[mask], axis=0)
            centroids.append(centroid)
    
    # Calculate distances between centroids
    if len(centroids) > 1:
        distances = cdist(centroids, centroids)
        np.fill_diagonal(distances, np.inf)  # Ignore self-distances
        
        min_distances = np.min(distances, axis=1)
        avg_min_distance = np.mean(min_distances)
        
        print(f"Average minimum inter-class distance: {avg_min_distance:.3f}")
        
        # Find closest and most distant classes
        min_idx = np.unravel_index(np.argmin(distances), distances.shape)
        max_idx = np.unravel_index(np.argmax(distances), distances.shape)
        
        print(f"Most similar classes: {class_names[min_idx[0]]} ↔ {class_names[min_idx[1]]}")
        print(f"Most different classes: {class_names[max_idx[0]]} ↔ {class_names[max_idx[1]]}")

def visualize_outliers(embeddings, labels, filenames, class_names):
    """Identify and visualize outliers in the feature space"""
    print("\n=== OUTLIER ANALYSIS ===")
    
    # Calculate distances from class centroids
    outliers = []
    
    for class_idx in range(len(class_names)):
        mask = labels == class_idx
        if np.sum(mask) > 1:  # Need at least 2 points for centroid
            class_embeddings = embeddings[mask]
            class_filenames = [filenames[i] for i in range(len(filenames)) if labels[i] == class_idx]
            
            centroid = np.mean(class_embeddings, axis=0)
            distances = np.linalg.norm(class_embeddings - centroid, axis=1)
            
            # Find points that are far from centroid (outliers)
            threshold = np.mean(distances) + 2 * np.std(distances)
            outlier_indices = np.where(distances > threshold)[0]
            
            for idx in outlier_indices:
                outliers.append({
                    'filename': class_filenames[idx],
                    'class': class_names[class_idx],
                    'distance': distances[idx],
                    'embedding': class_embeddings[idx]
                })
    
    if outliers:
        print(f"Found {len(outliers)} potential outliers:")
        for outlier in sorted(outliers, key=lambda x: x['distance'], reverse=True)[:5]:  # Top 5
            print(f"  {outlier['filename']} ({outlier['class']}): distance {outlier['distance']:.3f}")
    else:
        print("No significant outliers found.")

def compare_feature_spaces(features, labels, class_names):
    """Compare t-SNE and PCA side by side"""
    print("Creating comparison of feature space visualizations...")
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))
    
    # t-SNE
    tsne = TSNE(n_components=2, random_state=42, perplexity=min(30, len(features)-1))
    embeddings_tsne = tsne.fit_transform(features)
    
    # PCA
    pca = PCA(n_components=2, random_state=42)
    embeddings_pca = pca.fit_transform(features)
    variance_ratio = pca.explained_variance_ratio_
    
    colors = ['blue', 'red', 'green', 'orange', 'purple']
    
    # Plot t-SNE
    for class_idx in range(len(class_names)):
        mask = labels == class_idx
        ax1.scatter(embeddings_tsne[mask, 0], embeddings_tsne[mask, 1],
                   c=colors[class_idx % len(colors)], 
                   label=class_names[class_idx],
                   alpha=0.7, s=50)
    
    ax1.set_title('t-SNE Visualization\n(Non-linear dimensionality reduction)', fontsize=20, fontweight='bold')
    ax1.set_xlabel('t-SNE Component 1',fontsize=15)
    ax1.set_ylabel('t-SNE Component 2',fontsize=15)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Plot PCA
    for class_idx in range(len(class_names)):
        mask = labels == class_idx
        ax2.scatter(embeddings_pca[mask, 0], embeddings_pca[mask, 1],
                   c=colors[class_idx % len(colors)], 
                   label=class_names[class_idx],
                   alpha=0.7, s=50)
    
    ax2.set_title(f'PCA Visualization\n(Linear dimensionality reduction)\nExplained Variance: {variance_ratio[0]:.2f} + {variance_ratio[1]:.2f} = {variance_ratio.sum():.2f}', 
                 fontsize=20, fontweight='bold')
    ax2.set_xlabel('Principal Component 1',fontsize=15)
    ax2.set_ylabel('Principal Component 2',fontsize=15)
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    return embeddings_tsne, embeddings_pca

# Main execution
if __name__ == "__main__":
    # Your data paths
    normal_folder = r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\tzh_normalimags"
    defect_folders = [
        r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\tzh_wear\abrasive",
        r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\tzh_wear\adhasive", 
        r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\tzh_wear\cracks",
        r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\tzh_wear\delamination"      
    ]
    
    CLASS_NAMES = ['normal', 'abrasive', 'adhesive', 'cracks', 'delamination']
    MODEL_PATH = r"C:\Users\Administrator\OneDrive\Desktop\Gan+Unet project\hybrid codes\final_hybrid_model.pth"
    
    # Combine all folders
    all_folders = [normal_folder] + defect_folders
    
    print("Loading discriminator for feature extraction...")
    discriminator = load_discriminator(MODEL_PATH)
    
    print("Extracting features from your dataset...")
    features, labels, filenames = extract_features_from_images(
        discriminator, all_folders, CLASS_NAMES, num_images_per_class=30
    )
    
    print(f"Extracted {len(features)} samples with {features.shape[1]} features each")
    
    # Compare both visualization methods
    print("\n=== COMPARING FEATURE SPACES ===")
    embeddings_tsne, embeddings_pca = compare_feature_spaces(features, labels, CLASS_NAMES)

    
    # Outlier analysis
    visualize_outliers(embeddings_tsne, labels, filenames, CLASS_NAMES)
```
